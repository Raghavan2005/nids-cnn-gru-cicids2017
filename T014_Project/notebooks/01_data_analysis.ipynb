```python
# Count labels across all CICIDS2017 CSV files

label_counts = {}

for file in csv_files:
    print(f"Processing: {file.name}")

    df_temp = pd.read_csv(file)

    # Handle possible whitespace in the label column
    label_col = "Label" if "Label" in df_temp.columns else " Label"

    counts = df_temp[label_col].value_counts()

    for label, count in counts.items():
        label_counts[label] = label_counts.get(label, 0) + count

# Convert to a DataFrame
label_distribution = (
    pd.DataFrame.from_dict(
        label_counts,
        orient="index",
        columns=["Count"]
    )
    .sort_values("Count", ascending=False)
)

print("\nComplete label distribution:")
display(label_distribution)
```

```python
# Check NaN and infinite values in each CSV

data_quality = []

for file in csv_files:
    print(f"Checking: {file.name}")

    df_temp = pd.read_csv(file)

    # Strip whitespace from column names
    df_temp.columns = df_temp.columns.str.strip()

    # Numeric columns only
    numeric_cols = df_temp.select_dtypes(include=np.number).columns

    nan_count = df_temp[numeric_cols].isna().sum().sum()

    inf_count = np.isinf(
        df_temp[numeric_cols].to_numpy()
    ).sum()

    data_quality.append({
        "File": file.name,
        "Rows": len(df_temp),
        "NaN Values": nan_count,
        "Infinite Values": inf_count
    })

data_quality_df = pd.DataFrame(data_quality)

display(data_quality_df)
```

```python
print("Total NaN values:",
      data_quality_df["NaN Values"].sum())

print("Total infinite values:",
      data_quality_df["Infinite Values"].sum())
```

```python
# Find columns containing NaN or infinite values

problematic_columns = []

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    numeric_cols = df_temp.select_dtypes(include=np.number).columns

    nan_counts = df_temp[numeric_cols].isna().sum()
    inf_counts = pd.Series(
        np.isinf(df_temp[numeric_cols].to_numpy()).sum(axis=0),
        index=numeric_cols
    )

    for column in numeric_cols:
        if nan_counts[column] > 0 or inf_counts[column] > 0:
            problematic_columns.append({
                "File": file.name,
                "Column": column,
                "NaN": int(nan_counts[column]),
                "Infinity": int(inf_counts[column])
            })

problematic_df = pd.DataFrame(problematic_columns)

display(problematic_df)
```

```python
# Inspect rows containing infinite values

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    numeric_cols = df_temp.select_dtypes(include=np.number).columns

    inf_mask = np.isinf(df_temp[numeric_cols]).any(axis=1)

    if inf_mask.sum() > 0:
        print("\n" + "=" * 80)
        print(file.name)
        print(f"Rows containing infinity: {inf_mask.sum()}")

        display(
            df_temp.loc[
                inf_mask,
                ["Flow Bytes/s", "Flow Packets/s", "Label"]
            ].head(10)
        )
```

```python
# Count all rows containing NaN or infinity, grouped by label

problematic_label_counts = {}

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    numeric_cols = df_temp.select_dtypes(include=np.number).columns

    # Rows containing NaN or infinity in any numeric feature
    nan_mask = df_temp[numeric_cols].isna().any(axis=1)
    inf_mask = np.isinf(df_temp[numeric_cols].to_numpy()).any(axis=1)

    problematic_mask = nan_mask | inf_mask

    counts = df_temp.loc[problematic_mask, "Label"].value_counts()

    for label, count in counts.items():
        problematic_label_counts[label] = (
            problematic_label_counts.get(label, 0) + count
        )

problematic_by_label = (
    pd.DataFrame.from_dict(
        problematic_label_counts,
        orient="index",
        columns=["Problematic Rows"]
    )
    .sort_values("Problematic Rows", ascending=False)
)

display(problematic_by_label)
```

```python
# Calculate the impact of removing problematic rows
# for every label

impact_rows = []

for label in label_distribution.index:

    total = label_distribution.loc[label, "Count"]
    problematic = problematic_by_label.loc[label, "Problematic Rows"] \
        if label in problematic_by_label.index else 0

    remaining = total - problematic
    percentage_removed = (problematic / total) * 100

    impact_rows.append({
        "Label": label,
        "Original Count": total,
        "Problematic Rows": problematic,
        "Remaining Rows": remaining,
        "Removed (%)": percentage_removed
    })

cleaning_impact = pd.DataFrame(impact_rows)

display(
    cleaning_impact.sort_values(
        "Original Count",
        ascending=False
    )
)
```

```python
# Check duplicate rows in each CSV file

duplicate_results = []

for file in csv_files:
    print(f"Checking: {file.name}")

    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    total_rows = len(df_temp)

    duplicate_rows = df_temp.duplicated().sum()

    duplicate_results.append({
        "File": file.name,
        "Total Rows": total_rows,
        "Duplicate Rows": int(duplicate_rows),
        "Duplicate (%)": (duplicate_rows / total_rows) * 100
    })

duplicates_df = pd.DataFrame(duplicate_results)

display(duplicates_df)
```

```python
# Check whether duplicate feature vectors have conflicting labels

duplicate_label_analysis = []

for file in csv_files:
    print(f"Checking: {file.name}")

    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    # Separate features and label
    feature_cols = [col for col in df_temp.columns if col != "Label"]

    # Find duplicated feature vectors
    duplicated_features = df_temp.duplicated(
        subset=feature_cols,
        keep=False
    )

    duplicate_df = df_temp.loc[duplicated_features]

    if len(duplicate_df) > 0:
        # Number of unique labels associated with each feature vector
        label_counts_per_feature = (
            duplicate_df
            .groupby(feature_cols, dropna=False)["Label"]
            .nunique()
        )

        conflicting_groups = (
            label_counts_per_feature > 1
        ).sum()

        duplicate_groups = len(label_counts_per_feature)

    else:
        conflicting_groups = 0
        duplicate_groups = 0

    duplicate_label_analysis.append({
        "File": file.name,
        "Duplicate Feature Groups": int(duplicate_groups),
        "Conflicting Label Groups": int(conflicting_groups)
    })

duplicate_label_df = pd.DataFrame(duplicate_label_analysis)

display(duplicate_label_df)
```

```python
# Inspect conflicting duplicate feature groups in Wednesday

wednesday_file = [
    f for f in csv_files
    if "Wednesday" in f.name
][0]

df_wed = pd.read_csv(wednesday_file)
df_wed.columns = df_wed.columns.str.strip()

feature_cols = [
    col for col in df_wed.columns
    if col != "Label"
]

# Find feature vectors that have more than one label
label_counts = (
    df_wed
    .groupby(feature_cols, dropna=False)["Label"]
    .nunique()
)

conflicting_features = label_counts[
    label_counts > 1
].index

conflicting_rows = (
    df_wed
    .set_index(feature_cols)
    .loc[conflicting_features]
    .reset_index()
)

print("Number of conflicting feature groups:",
      len(conflicting_features))

print("\nRows involved:")
print(len(conflicting_rows))

display(
    conflicting_rows[
        feature_cols + ["Label"]
    ].head(100)
)
```

```python
# Inspect conflicting duplicate feature groups in Wednesday

wednesday_file = [
    f for f in csv_files
    if "Wednesday" in f.name
][0]

df_wed = pd.read_csv(wednesday_file)
df_wed.columns = df_wed.columns.str.strip()

feature_cols = [
    col for col in df_wed.columns
    if col != "Label"
]

# Find feature vectors that occur with more than one label
label_counts = (
    df_wed
    .groupby(feature_cols, dropna=False)["Label"]
    .nunique()
)

# Identify conflicting feature groups
conflicting_features = label_counts[label_counts > 1]

print("Number of conflicting feature groups:",
      len(conflicting_features))

# Create a mask for rows belonging to conflicting groups
# using a merge instead of MultiIndex .loc
conflict_keys = conflicting_features.reset_index()[feature_cols]

conflicting_rows = df_wed.merge(
    conflict_keys,
    on=feature_cols,
    how="inner"
)

print("Rows involved in conflicts:",
      len(conflicting_rows))

print("\nLabels involved:")
display(
    conflicting_rows["Label"]
    .value_counts()
)

print("\nConflicting rows:")
display(
    conflicting_rows[
        feature_cols + ["Label"]
    ].head(100)
)
```

```python
# Show which labels conflict with each other

conflict_pairs = set()

for _, group in conflicting_rows.groupby(feature_cols, dropna=False):
    labels = sorted(group["Label"].unique())

    if len(labels) > 1:
        for i in range(len(labels)):
            for j in range(i + 1, len(labels)):
                conflict_pairs.add(
                    (labels[i], labels[j])
                )

print("Conflicting label pairs:\n")

for pair in sorted(conflict_pairs):
    print(f"{pair[0]}  <-->  {pair[1]}")
```

```python
# Calculate the exact impact of removing conflicting
# feature-vector groups from Wednesday

conflict_row_count = len(conflicting_rows)

print("Rows involved in conflicting groups:", conflict_row_count)

print("\nRows removed by label:")
display(
    conflicting_rows["Label"].value_counts()
)

# Compare against Wednesday's original class distribution
wednesday_labels = (
    df_wed["Label"]
    .value_counts()
    .rename("Original Count")
    .to_frame()
)

removed_labels = (
    conflicting_rows["Label"]
    .value_counts()
    .rename("Conflicting Rows")
    .to_frame()
)

conflict_impact = (
    wednesday_labels
    .join(removed_labels, how="left")
    .fillna(0)
)

conflict_impact["Remaining"] = (
    conflict_impact["Original Count"]
    - conflict_impact["Conflicting Rows"]
)

conflict_impact["Removed (%)"] = (
    conflict_impact["Conflicting Rows"]
    / conflict_impact["Original Count"]
    * 100
)

display(conflict_impact)
```

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path(
    r"D:\PS26-PAPER1\data\processed\cicids2017_clean_selected.csv"
)

OUTPUT_PATH = Path(
    r"D:\PS26-PAPER1\data\processed\cicids2017_final.csv"
)

df = pd.read_csv(DATA_PATH)

print("=" * 70)
print("GLOBAL CONFLICT CLEANING")
print("=" * 70)

print("Original shape:", df.shape)

feature_cols = [
    col for col in df.columns
    if col != "Label"
]

# ------------------------------------------------------------
# Find feature vectors associated with multiple labels
# ------------------------------------------------------------

label_counts = (
    df.groupby(
        feature_cols,
        dropna=False
    )["Label"]
    .transform("nunique")
)

conflict_mask = label_counts > 1

conflicting_rows = conflict_mask.sum()

print(
    "\nRows belonging to conflicting groups:",
    conflicting_rows
)

# ------------------------------------------------------------
# Remove ALL rows belonging to conflicting groups
# ------------------------------------------------------------

df_final = df.loc[
    ~conflict_mask
].copy()

print(
    "Rows after conflict removal:",
    len(df_final)
)

print(
    "Rows removed:",
    len(df) - len(df_final)
)

# ------------------------------------------------------------
# Verify no conflicts remain
# ------------------------------------------------------------

remaining_conflicts = (
    df_final
    .groupby(
        feature_cols,
        dropna=False
    )["Label"]
    .nunique()
)

remaining_conflicts = (
    remaining_conflicts > 1
).sum()

print(
    "\nRemaining conflicting feature groups:",
    remaining_conflicts
)

# ------------------------------------------------------------
# Verify NaN / Inf
# ------------------------------------------------------------

numeric_cols = df_final.select_dtypes(
    include="number"
).columns

nan_count = df_final[numeric_cols].isna().sum().sum()

inf_count = (
    df_final[numeric_cols]
    .isin([float("inf"), float("-inf")])
    .sum()
    .sum()
)

print("Remaining NaN values:", nan_count)
print("Remaining infinite values:", inf_count)

# ------------------------------------------------------------
# Class distribution
# ------------------------------------------------------------

print("\nFinal class distribution:")

print(
    df_final["Label"].value_counts()
)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

df_final.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\nSaved final dataset to:")
print(OUTPUT_PATH)
```
